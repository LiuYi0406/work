## 实验2：函数、条件、循环与递归

```python
# ========= 任务1 score_level成绩等级函数 =========
def score_level(score):
    if not (0 <= score <= 100):
        raise ValueError("分数必须在0~100之间")
    if 90 <= score <= 100:
        return "优秀"
    elif 80 <= score < 90:
        return "良好"
    elif 60 <= score < 80:
        return "及格"
    else:
        return "不及格"

print("score_level(95):", score_level(95))
print("score_level(85):", score_level(85))
print("score_level(70):", score_level(70))
print("score_level(50):", score_level(50))
try:
    print(score_level(105))
except ValueError as e:
    print("异常测试：", e)
```

```python
# ========= 任务2 for、while求1‑100和 =========
# for循环
sum_for = 0
for i in range(1, 101):
    sum_for += i
print("for循环1‑100和：", sum_for)

# while循环
sum_while = 0
i = 1
while i <= 100:
    sum_while += i
    i += 1
print("while循环1‑100和：", sum_while)
```

- for：直接遍历`range(1,101)`累加；  
- while：设置计数器 i，i≤100 循环累加，i 自增；  
两种结果相等，1~100 和等于 5050。  

```python
# ========= 任务3 迭代版、递归版阶乘 =========
# 迭代版
def fact_iter(n):
    if n < 0:
        raise ValueError("n应为非负整数")
    res = 1
    for i in range(2, n+1):
        res *= i
    return res

# 递归版
def fact_recur(n):
    if n < 0:
        raise ValueError("n应为非负整数")
    if n == 0 or n == 1:
        return 1
    return n * fact_recur(n-1)

for n in [0,1,5,10]:
    print(f"n={n},迭代:{fact_iter(n)},递归:{fact_recur(n)}")
```

- 迭代阶乘：循环从 2 到 n 累乘；  
- 递归阶乘：基线条件`n==0或n==1返回1`，递归式 `n! = n(n‑1)！`  
测试 n=0,1,5,10，迭代与递归结果完全一致。  

```python
# ========= 任务4 summary 返回(最小,最大,均值)，元组解包 =========
def summary(values):
    if not values:
        raise ValueError("列表不能为空")
    min_v = min(values)
    max_v = max(values)
    avg_v = sum(values)/len(values)
    return (min_v, max_v, avg_v)

lst = [12,45,7,89,23]
min_num, max_num, avg_num = summary(lst)   #元组解包
print("summary测试，列表",lst)
print(f"最小值:{min_num},最大值:{max_num},均值:{avg_num:.2f}")
```

```python
# ========= 任务5 sum_all(*args)可变参数求和 =========
def sum_all(*args):
    return sum(args)

print("无参数 sum_all():", sum_all())
print("单参数 sum_all(10):", sum_all(10))
print("多参数 sum_all(1,2,3,4,5):", sum_all(1,2,3,4,5))
```

`*args`接收可变数量位置参数，把全部参数打包成元组；`sum(args)`求和。  
- 无参数：args 为空元组，返回 0  
- 单个参数：对一个数求和  
- 多个参数：全部参数累加

## 实验4：文件读写、异常处理与模块

```python
import pickle

# ========= 任务1 创建目录 =========
cwd = os.getcwd()   # 获取当前工作目录
print("当前工作目录：", cwd)
data_dir = os.path.join(cwd, "exp01_data")
os.makedirs(data_dir, exist_ok=True)
print("数据目录路径：", data_dir)
```

```python
# ========= 任务2 写入students.txt，再读取，跳过表头 =========
txt_path = os.path.join(data_dir, "students.txt")

content_text = """name,score
Alice,85
Bob,58
Carol,92
David,76
Eva,60
Frank,89
Grace,abc
Henry,105
Incomplete
"""

# with写入文件
with open(txt_path, "w", encoding="utf-8") as f:
    f.write(content_text)

# 读取，跳过第一行表头
lines_all = []
with open(txt_path, "r", encoding="utf-8") as f:
    next(f)   #跳过表头
    for line in f:
        lines_all.append(line.rstrip("\n"))

for l in lines_all:
    print(l)
```

```python
# ========= 任务3 解析函数 parse_student =========
def parse_student(line):
    line = line.strip()
    parts = line.split(",")
    #检查字段数量
    if len(parts) != 2:
        raise ValueError(f"字段数量错误，得到{len(parts)}个字段")
    name_str, score_str = parts
    name_str = name_str.strip()
    score_str = score_str.strip()
    #姓名非空
    if not name_str:
        raise ValueError("姓名为空")
    #成绩转整数
    try:
        score_int = int(score_str)
    except ValueError:
        raise ValueError(f"成绩无法转为整数：{score_str}")
    #成绩范围0‑100
    if not (0 <= score_int <= 100):
        raise ValueError(f"分数{score_int}超出0‑100范围")
    return {"name": name_str, "score": score_int}
```

```python
# ========= 任务4 逐行处理，捕获异常，保存有效记录 =========
valid_records = []
error_log = []

for row_idx, one_line in enumerate(lines_all, start=2): #文件行号，表头是第1行
    try:
        record = parse_student(one_line)
        valid_records.append(record)
    except ValueError as err:
        error_log.append({"line": row_idx, "reason": str(err)})

print("有效记录：")
for r in valid_records:
    print(r)
print("\n错误记录日志：")
for e in error_log:
    print(e)
```

```python
# ========= 任务5 写入report.txt；演示 w模式 和 a模式 =========
valid_cnt = len(valid_records)
invalid_cnt = len(error_log)
if valid_cnt > 0:
    avg_score = sum(r["score"] for r in valid_records)/valid_cnt
    pass_cnt = sum(1 for r in valid_records if r["score"]>=60)
    pass_rate = pass_cnt / valid_cnt
else:
    avg_score = 0
    pass_rate = 0

report_path = os.path.join(data_dir, "report.txt")

# w模式：覆盖原有全部内容
with open(report_path, "w", encoding="utf‑8") as f:
    f.write(f"有效记录数：{valid_cnt}\n")
    f.write(f"无效记录数：{invalid_cnt}\n")
    f.write(f"平均分：{avg_score:.2f}\n")
    f.write(f"及格率：{pass_rate:.2%}\n")

# a追加模式：在文件末尾新增文字，不删除旧内容
with open(report_path, "a", encoding="utf‑8") as f:
    f.write("===这是a模式追加的一行文字===\n")

print(f"report.txt已经生成在 {report_path}")
```

- `w`写模式：打开文件直接清空原有全部内容，写入新内容。  
- `a`追加模式：不删除原有内容，新内容写在文件末尾。  

```python
# ========= 任务6 pickle序列化保存与读取 =========
pkl_path = os.path.join(data_dir, "students.pkl")
# dump保存
with open(pkl_path, "wb") as f:
    pickle.dump(valid_records, f)

# load读取
with open(pkl_path, "rb") as f:
    loaded_data = pickle.load(f)

print("pickle读取后数据是否与原数据相等：", loaded_data == valid_records)
```

- `pickle.dump(对象,文件对象)`：把 Python 对象序列化保存二进制文件；  
- `pickle.load(文件对象)`：从二进制文件恢复 Python 对象；对比恢复对象与原对象是否完全相等。

##  实验5:类、继承与对象复制

```python
import copy

# ========== 任务1：定义Student类 ==========
class Student:
    def __init__(self, name, score):
        self.name = name
        self.score = score

    def is_passed(self):
        """判断是否及格，>=60返回True"""
        return self.score >= 60

    def __str__(self):
        return f"Student(name={self.name}, score={self.score}, passed={self.is_passed()})"
```

- `__init__`构造方法初始化实例属性`name`、`score`  
- `is_passed()`实例方法，判断分数≥60 是否及格  
- `__str__`：print 对象时自动调用，返回对象可读字符串。  

```python
# ========== 任务2：用实验4 valid_records创建Student对象 ==========
student_obj_list = []
for item in valid_records:
    s = Student(item["name"], item["score"])
    student_obj_list.append(s)
    print(s)   #调用__str__，输出对象信息和及格状态
```

```python
# ========== 任务3：子类GraduateStudent继承Student ==========
class GraduateStudent(Student):
    def __init__(self, name, score, research_topic):
        super().__init__(name, score)   #调用父类构造
        self.research_topic = research_topic

    def __str__(self):
        return f"GraduateStudent(name={self.name}, score={self.score}, research_topic={self.research_topic}, passed={self.is_passed()})"

#测试子类
g_stu = GraduateStudent("Tom",78,"大数据分析")
print(g_stu)
```

```python
# ========== 任务4：直接赋值、浅拷贝copy.copy、深拷贝copy.deepcopy ==========
#原始嵌套对象
original = {"name": "Alice", "scores": [85, 90]}

# 1. 直接赋值
obj1 = original
obj1["scores"][0] = 100
print(f"直接赋值后：original={original}, obj1={obj1}")
#恢复原始
original = {"name": "Alice", "scores": [85, 90]}

# 2. 浅拷贝 copy.copy()
obj2 = copy.copy(original)
obj2["scores"][0] = 100
print(f"浅拷贝copy.copy后：original={original}, obj2={obj2}")
original = {"name": "Alice", "scores": [85, 90]}

# 3. 深拷贝 copy.deepcopy()
obj3 = copy.deepcopy(original)
obj3["scores"][0] = 100
print(f"深拷贝copy.deepcopy后：original={original}, obj3={obj3}")
original = {"name": "Alice", "scores": [85, 90]}
```

- 直接赋值：只是同一个对象别名，完全共享，改一个全部跟着变。  
- copy.copy () 浅拷贝：外层新建容器，嵌套可变对象（列表）仍然共用引用；修改内部列表，原始对象同步改变。  
- copy.deepcopy () 深拷贝：递归复制所有嵌套对象，完全独立副本；修改副本内部列表，原始对象不受影响。
