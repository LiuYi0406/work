## 欧几里得算法

**原理**：  
对于整数 m,n：
`gcd(m,n) = gcd(n, m mod n)`，重复取模，直到余数`n==0`，此时 m 就是最大公约数。

```python
# ========== 1. 实现gcd函数 ==========
def gcd(m, n):
    while n != 0:
        m, n = n, m % n
    return m


# ========== 2. 测试三组数据 ==========
print(gcd(2018, 2019))
print(gcd(1200, 300))
print(gcd(5040, 60))
```

```python
# ========== 3. 测试负数参数 ==========
print(gcd(-12, -8))
print(gcd(-12, 8))
```

当传入负数参数，函数依然可以完成循环计算，但返回结果有可能是负数。  
最大公约数在数学定义中只针对非负整数，公约数一定为正数，负数的输出没有实际数学意义。  
改进方案：可以先对两个输入取绝对值，`m = abs(m); n = abs(n)`再做计算，保证输出为正的最大公约数。  

```python
# ========== 4. 改进函数gcd，可处理负数参数 ==========
def gcd_fixed(m,n):
    m = abs(m)
    n = abs(n)
    while n != 0:
        m,n = n, m%n
    return m

# 测试
print(gcd_fixed(-12, -8))
print(gcd_fixed(-12, 8))
print(gcd_fixed(2018, 2019))
```

经过测试验证，改进后的gcd_fixed函数能正确处理传入参数为负的情况。

## 近似欧拉数e

欧拉数 e 有两种表达方式：
- 极限公式：$ e=\lim_{x\to\infty}\left(1+\frac1x\right)^x $，x越大越接近 e。
- 阶乘无穷级数：$ e=\sum_{k=0}^{\infty}\frac{1}{k!} $。

```python
# ========== 1. 极限公式 ==========
# 用极限公式 (1+1/n)^n 近似e
def euler_limit(n):
    return (1 + 1 / n) ** n

# 测试
print(euler_limit(10000))
```

```python
# ========== 2. 阶乘无穷级数 ==========
# 前n项求和，公式 e = sum(1/k!)，k从0开始
def euler_infinite_sum(n):
    if n == 0:
        return 0.0
    total = 0.0
    fact = 1   # 0! = 1 
    for k in range(n):
        total += 1.0 / fact
        fact *= (k + 1)
    return total

# 测试
print(euler_infinite_sum(0))   
print(euler_infinite_sum(1))  
print(euler_infinite_sum(4))   
```

```python
import math

# ========== 3. 自适应精度求e ==========
# 循环累加，直到和与真实e的误差小于epsilon
def euler_approx(epsilon):
    e_real = math.exp(1)  # 真实 e 取math.exp(1)
    total = 0.0
    fact = 1
    k = 0
    while abs(total - e_real) > epsilon:
        total += 1.0 / fact
        k += 1
        fact *= k
    return total

# 测试
print(euler_approx(0.001))
```

```python
# ========== 4. 输出级数收敛表格 ==========
# 打印k从1~n的euler_infinite_sum(k)
def print_euler_sum_table(n):
    for k in range(1, n + 1):
        print(euler_infinite_sum(k))

# 测试
print_euler_sum_table(5)
```

两种方式可以近似计算欧拉数e：极限形式$ e\approx\left(1+\frac1n\right)^n $和阶乘级数求和。  
随着项数n增大，两种方法得到的近似值都会不断逼近 e 的真实值。  
级数求和收敛速度明显更快，只需要很少项就可以达到很高精度；极限公式则需要取非常大的n才能得到相近精度。  
通过设置误差阈值ε，可以自动控制计算，得到满足精度要求的近似结果。

## 测试整数属性

**数学原理：**  
n 如果是 b 的 p 次方，n 不断除以 b，最后会得到 1。  
边界：$p=0$，$b^0=1$，只要 n=1，不管 b 是几，直接 True。  
循环逻辑：只要 `n % b == 0`，代表还能被 b 整除，就执行 `n = n // b`。  
循环结束后看剩下的值：  

- 剩余等于 1：$n=b^p$ → True
- 剩余不等于 1：含有别的因子 → False

```python
# ========== 1. 判断正整数n是否为2的幂 ==========
def is_power_of_2(n):
    # 不断除以2，直到不能被2整除
    while n % 2 == 0:
        n = n // 2
    # 最后结果等于1，说明是2的幂：2^p
    return n == 1

# 测试
print(is_power_of_2(1))
print(is_power_of_2(8))
print(is_power_of_2(10))
```

```python
# ========== 2. 判断 n 是否是 b 的幂 ==========
def is_power(b, n):
    # 特殊：任何数的0次幂等于1，n=1永远返回True
    if n == 1:
        return True
    # 不断把n除以b，直到不能整除
    while n % b == 0:
        n = n // b
    # 最后剩下1，代表可以写成b的p次方
    return n == 1

# 测试
print(is_power(3, 9))   
print(is_power(5, 125))
print(is_power(2, 10))  
print(is_power(7,1))   
```
