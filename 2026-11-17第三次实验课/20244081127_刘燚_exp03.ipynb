##  实验1：正则表达式的基础配置

```python
import re

# ========== 任务1 ==========
text1 = "cat on mat; later: cats and dogs"

res1_1 = re.findall(r"\d+", text1)   
res1_2 = re.findall(r"\s+", text1)    
res1_3 = re.findall(r"\bcat\b", text1)

print(r"\d+ 匹配结果：", res1_1)
print(r"\s+ 匹配结果：", res1_2)
print(r"\bcat\b 匹配结果：", res1_3)
```

- `\d+`：匹配 1 个及以上数字，原文无数字 → 空列表
- `\s+`：匹配连续空白（空格、换行等）
- `\bcat\b`：单词边界，只匹配独立单词`cat`，**不会匹配 cats**

```python
# ========== 任务2 ==========
text2 = "cat dog bird CAT"

res2_1 = re.findall(r"cat|dog", text2)
res2_2 = re.findall(r"a|aa|aaa", "aaaa")
res2_3 = re.findall(r"aaa|aa|a", "aaaa")

print(r"cat|dog 匹配结果：", res2_1)
print(r"a|aa|aaa 匹配'aaaa'：", res2_2)
print(r"aaa|aa|a 匹配'aaaa'：", res2_3)
```

`|` 是**或**，正则是**从左向右优先匹配**：

- `a|aa|aaa`：先匹配短的`a`，逐个拿字符，得到 4 个`a`
- `aaa|aa|a`：先尝试匹配最长`aaa`，剩下 1 个`a`，得到`['aaa','a']`

```python
# ========== 任务3 ==========
text3 = "ct cat caat caaat"

res3_1 = re.findall(r"ca*t", text3)
res3_2 = re.findall(r"ca+t", text3)
res3_3 = re.findall(r"ca?t", text3)
res3_4 = re.findall(r"ca{2}t", text3)
res3_5 = re.findall(r"ca{1,2}t", text3)

print(r"ca*t ：", res3_1)
print(r"ca+t ：", res3_2)
print(r"ca?t ：", res3_3)
print(r"ca{2}t：", res3_4)
print(r"ca{1,2}t：", res3_5)
```

| 量词 | 含义 |
| ---- | ---- |
| `ca*t` | `*`：前面字符 a 出现 0 次或多次 |
| `ca+t` | `+`：前面字符 a 出现 1 次或多次 |
| `ca?t` | `?`：前面字符 a 出现 0 次或 1 次 |
| `ca{2}t` | `{n}`：前面字符 a 恰好 2 次 |
| `ca{1,2}t` | `{m,n}`：前面字符 a 出现 1~2 次（包含边界） |


## 实验2：分组、回溯引用与替换

```python
import re

# ========== 任务1：捕获组提取邮箱用户名、域名 ==========
text1 = "联系：alice@example.com ; 备选：bob@test.org"

# 分组：第一组用户名@前面，第二组域名@后面
pattern1 = r"(\w+)@([\w.]+)"
res1 = re.findall(pattern1, text1)

print("用户名,域名：", res1)
```

```python
# ========== 任务2：后向引用匹配连续重复单词 ==========
text2 = "cat cat, dog cat, bird bird"

# \b 单词边界；(\w+)捕获单词；\1 引用第1分组，匹配相同内容
pattern2 = r"\b(\w+)\s+\1\b"
res2 = re.findall(pattern2, text2)

print("连续重复单词：", res2)
```

```python
# ========== 任务3：日期格式转换 年-月-日 → 月-日-年 ==========
text3 = "2024-05-01\n2024/05/02\n2024.05.03"

# 分隔符可以是 - / .
pattern3 = r"(\d{4})[-/.](\d{2})[-/.](\d{2})"
# 替换：\2-\3-\1
text3_new = re.sub(pattern3, r"\2-\3-\1", text3)

print("转换前：")
print(text3)
print("转换后：")
print(text3_new)
```

```python
# ========== 任务4：电话号码统一格式 (123) 456-7890 ==========
text4 = "联系我：123-456-7890 或 987.654.3210"

pattern4 = r"(\d{3})[-.](\d{3})[-.](\d{4})"
text4_new = re.sub(pattern4, r"(\1) \2-\3", text4)

print("转换前：", text4)
print("转换后：", text4_new)
```

## 实验3：使用 BeautifulSoup 解析 HTML

```python
from bs4 import BeautifulSoup

html_doc = """
<html><head><title>实验课程</title></head><body>
<h1>数据分析</h1>
<p class="course">第六讲：正则表达式</p>
<p class="course">第七讲：结构化数据</p>
<a href="https://example.org/lecture06">讲义六</a>
<a href="https://example.org/lecture07">讲义七</a>
</body></html>
"""

# ========== 任务1：构建解析树，输出标题文本、h1标签名与文本 ==========
soup = BeautifulSoup(html_doc, "html.parser")
title_tag = soup.title
h1_tag = soup.h1

print("页面标题文本：", title_tag.get_text())
print("h1标签名：", h1_tag.name)
print("h1文本：", h1_tag.get_text())
```

- `.name`：获取标签名称；
- `.get_text()`：提取标签内部纯文本

```python
# ========== 任务2：find_all找class=course的p标签、所有a链接，输出链接文字和href ==========
p_list = soup.find_all("p", class_="course")
print("所有class='course'的段落：")
for p in p_list:
    print(p.get_text())

a_list = soup.find_all("a")
print("\n所有链接：")
for a in a_list:
    link_text = a.get_text()
    link_href = a["href"]
    print(f"链接文字：{link_text}，href：{link_href}")
```

```python
# ========== 任务3：get_text()提取文本，对比文档、单个段落、单个链接 ==========
print("整个文档get_text():")
print(soup.get_text())

print("\n第一个p段落get_text():")
print(p_list[0].get_text())

print("\n第一个a链接get_text():")
print(a_list[0].get_text())
```
